```python
from tensorflow.keras.datasets import fashion_mnist
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. Carregar os dados (como você já fez)
(X_train, y_train), (X_test, y_test) = fashion_mnist.load_data()
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.optimizers import Adam

# Normalizar os dados
X_train_normalized = X_train / 255.0
X_test_normalized = X_test / 255.0

# Criar o modelo MLP
model = Sequential([
    Flatten(input_shape=(28, 28)),  # Achatar as imagens 28x28 em vetores 1D
    Dense(128, activation='relu'),  # Primeira camada oculta com 128 neurônios
    Dense(64, activation='relu'),   # Segunda camada oculta com 64 neurônios
    Dense(10, activation='softmax') # Camada de saída com 10 neurônios (uma para cada classe)
])

# Compilar o modelo
model.compile(optimizer=Adam(learning_rate=0.001),
            loss='sparse_categorical_crossentropy',
            metrics=['accuracy'])


test_loss, test_accuracy = model.evaluate(X_test_normalized, y_test)
print(f"Acurácia no conjunto de teste: {test_accuracy * 100:.2f}%")
# Avaliar o modelo no conjunto de teste

model.fit(X_train_normalized, y_train, epochs=50, batch_size=len(X_train_normalized), validation_split=0.2)
```

```python
evaluation = model.evaluate(X_test_normalized, y_test)
print(f"Loss: {evaluation[0]}, Accuracy: {evaluation[1]}")
```

```python
import os
lista_de_pesos = model.get_weights()
# Vamos separar em variáveis claras
W1, b1 = lista_de_pesos[0], lista_de_pesos[1]
W2, b2 = lista_de_pesos[2], lista_de_pesos[3]
W3, b3 = lista_de_pesos[4], lista_de_pesos[5]

# Verificar os shapes (formatos)
# Keras usa (input_dim, output_dim), vamos transpor para (output_dim, input_dim)
# para manter a consistência com a notação matemática (W.x + b)
W1, W2, W3 = W1.T, W2.T, W3.T
b1, b2, b3 = b1.reshape(-1, 1), b2.reshape(-1, 1), b3.reshape(-1, 1)

print("\\nShapes dos pesos extraídos e transpostos:")
print(f"W1 shape: {W1.shape}, b1 shape: {b1.shape}")
print(f"W2 shape: {W2.shape}, b2 shape: {b2.shape}")
print(f"W3 shape: {W3.shape}, b3 shape: {b3.shape}")

# --- 5. Salvar os Pesos em Arquivos ---
# Criar um diretório para salvar os pesos, se não existir
if not os.path.exists('pesos'):
    os.makedirs('pesos')

# Salvar cada array em um arquivo .npy separado
np.save('pesos/w1.npy', W1)
np.save('pesos/b1.npy', b1)
np.save('pesos/w2.npy', W2)
np.save('pesos/b2.npy', b2)
np.save('pesos/w3.npy', W3)
np.save('pesos/b3.npy', b3)

print("\\nPesos salvos com sucesso na pasta 'pesos/'!")
```
